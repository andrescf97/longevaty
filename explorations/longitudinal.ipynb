```python
import vital.metrics as metrics
import json
import pandas as pd
import numpy as np
```

```python
prediction_loc = "/nlst_data/checkpoints/longivity/predictions_vocal-shadow-44.json"
with open(prediction_loc) as fp:
    result = json.load(fp)

with open("/nlst_data/train_longi.json") as fp:
    monai_dict_train = json.load(fp)
```

```python
train_censoring_distribution = metrics.get_censoring_dist(monai_dict_train)
```

```python
df = pd.DataFrame(result)
```
